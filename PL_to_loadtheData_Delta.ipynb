```python
df = (
    spark.read.format("csv")
    .option("header","true")
    .option("inferSchema","true")
    .load("dbfs:/Volumes/main/default/my_volume/emp_Latest")
)

dept10_df = df.filter("dept_no = 10")

dept10_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("emp_dept10")
```
